```python
import pickle
import numpy as np
def read_sensors_from_sample(calibration_measure_path:str, sensor_names:list):
    with open(calibration_measure_path, "rb") as file:
        try:
            dict_data = pickle.load(file)
            if isinstance(dict_data, dict):
                output_dict = {}
                for sensor_name in sensor_names:
                    output_dict[sensor_name] = dict_data[sensor_name]
                return output_dict
            else:
                return Exception("Pickle returned something that's not a dictionary")
        except Exception as e:
            print(f"File Read Error: {e}")
            return e

def write_pickle_dict(calibration_args_path: str, dict_params: dict) -> None | Exception:
    with open(calibration_args_path, "wb") as file:
        try:
            pickle.dump(dict_params, file)
        except Exception as e:
            return e
```
